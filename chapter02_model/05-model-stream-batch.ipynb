流式调用（stream）

```python
from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek
import os

#加载配置文件
load_dotenv(override=True)

DEEPSEEK_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_URL = os.getenv("DEEPSEEK_BASE_URL")

#获取大模型
model = ChatDeepSeek(
    #model = "deepseek-v4-flash",
    model = "deepseek-v4-flash",
    api_key = DEEPSEEK_KEY,
    api_base = DEEPSEEK_URL
)

#print(model.invoke( "1+1 = ?"))
print(model.invoke( "你是谁？"))
```

```python
for chunk in model.stream("帮我解释一下什么是人工智能？"):
    print(chunk.text,end="",flush=True)
```

batch()批量调用

```python
messages = [
    "你好，你是谁？",
    "2 + 3 * 5 = ?",
    "中国首都在哪里？"
]

responses = model.batch(messages)

for response in responses:
    print(response.text,end="",flush=True)
```

按照完成的顺序接受响应

```python
messages = [
    "你好，你是谁？",
    "2 + 3 * 5 = ?",
    "中国首都在哪里？"
]

responses = model.batch_as_completed(messages)

for response in responses:
    print(response)
```
