```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, AIMessage
from langchain_ollama import ChatOllama
from mcp.server.fastmcp.prompts.base import UserMessage

model = init_chat_model(
    model="deepseek-r1:8b",
    base_url='http://localhost:11434',
    model_provider="ollama"
)

# print(model.invoke("介绍下你自己"))
```

```python
responses = model.invoke("翻译如下的汉字：你好世界")
print(responses)
```

```python
messages = [
    {"role":"system","content":"你是一个专业的数学老师"},
    {"role":"user","content":"帮我解释一下什么是斐波那契数列"}
]

responses = model.invoke(messages)
print(responses)
```

多轮对话

```python
messages = [
    {"role":"system","content":"你是一个专业的数学老师"},
    {"role":"user","content":"1+2=？"},
    {"role":"assistant","content":"3"},
    {"role":"user","content":"我刚才问了什么问题"}
]

responses = model.invoke(messages)
print(responses)
```

举例3：不传递历史，ai会失忆

```python
messages = [
    {"role":"system","content":"你是一个非常友好的AI助手"},
    {"role":"user","content":"你好，我叫小明"},
]

responses = model.invoke(messages)
print(f"AI的回复：{responses.content}")

messages2 = [
    {"role":"user","content":"我叫什么名字"},
]

responses2 = model.invoke(messages2)
print(responses2)
```

作为对比，传递记忆

```python
conversation = [
    {"role":"system","content":"你是一个非常友好的AI助手"},
    {"role":"user","content":"你好，我叫小明"},
]

responses = model.invoke(conversation)
print(f"AI的回复：{responses.content}")


#添加记忆
conversation.append({"role":"assistant","content":responses.content})
conversation.append({"role":"user","content":"我叫什么名字"})

#第二次对话
responses2 = model.invoke(conversation)
print(responses2)
```

消息对象列表

```python
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage

messages = [
    SystemMessage(content="你是一个专业的数学老师"),
    HumanMessage(content="帮我解释一下什么是斐波那契数列")
]

responses = model.invoke(messages)
print(responses)

#添加记忆
messages.append(AIMessage(content=responses.content))
messages.append(HumanMessage(content="我刚才问了什么问题"))

response1 = model.invoke(messages)
print(response1)
```
